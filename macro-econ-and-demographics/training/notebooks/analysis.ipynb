## 1. Pensionsfrågan

Pay-as-you-go-system bygger på ett enkelt förhållande: nuvarande arbetare betalar för nuvarande pensionärer. När förhållandet mellan arbetare och pensionärer kollapsar faller matematiken.

Tyskland pumpar ungefär en tredjedel av sina federala skatteinkomster in i den lagstadgade pensionsförsäkringen. Italiens INPS kräver ungefär 95 miljarder euro i årliga statsoverföringar för att täcka strukturella underskott. USA kan enligt lag inte ta allmänna skattemedel för Social Security — dess direkta subvention är noll, men dess förtroendefond beräknas ta slut i mitten av 2030-talet.

Varje variant av detta problem går tillbaka till samma rot: sjunkande fruktsamhet minskade poolen av framtida skattebetalare årtionden innan räkningarna förföll.

```python
from statsmodels.tsa.stattools import adfuller, grangercausalitytests

# =========================================================================
# 1. PREPARE THE ALIGNED DATASET
# =========================================================================
# Combining our long-term aligned variables into a clean testing DataFrame
# (Using the debt_final_long and social_final or fertility_final from your steps)
df_gc = pd.DataFrame({
    'Fertility': fertility_final,
    'Debt': debt_final
}).dropna()

print("--- STEG 1: KONTROLL AV STATIONARITET (ADF-TEST) ---")

# =========================================================================
# 2. THE STATIONARITY CHECK (AUGMENTED DICKEY-FULLER)
# =========================================================================
def check_stationarity(series, name):
    result = adfuller(series)
    p_value = result[1]
    print(f"{name} ADF p-value: {p_value:.4f}")
    if p_value < 0.05:
        print(f"  => {name} is Stationary (Ready for Granger Test).")
    else:
        print(f"  => {name} is Non-Stationary (Must be differenced).")
    return p_value

debt_p = check_stationarity(df_gc['Debt'], 'Debt')
fert_p = check_stationarity(df_gc['Fertility'], 'Fertility')

# =========================================================================
# 3. DIFFERENCING THE DATA (STRIPPING THE LONG-TERM TRENDS)
# =========================================================================
# If either series is non-stationary (which they likely are), we take the 1st difference.
if debt_p >= 0.05 or fert_p >= 0.05:
    print("\n--- STEP 2: DIFFERENCING TIME SERIES ---")
    df_gc_diff = df_gc.diff().dropna()
    
    # Re-verify stationarity of the differenced data
    check_stationarity(df_gc_diff['Debt'], 'Differentierad skuld')
    check_stationarity(df_gc_diff['Fertility'], 'Differentierad fruktsamhet')
else:
    df_gc_diff = df_gc

# =========================================================================
# 4. RUNNING THE GRANGER CAUSALITY TEST
# =========================================================================
print("\n--- STEP 3: RUNNING GRANGER CAUSALITY TEST ---")
# CRITICAL NOTE ON STATSMODELS: The function expects a 2D array/DataFrame where:
# - Column 1 is the DEPENDENT variable (the effect we want to predict)
# - Column 2 is the INDEPENDENT variable (the potential cause)
# We look across a 4-year lag window since demographic shifts take time to materialize.

max_lag = 4

print("\n>> Testing: Does DEBT Granger-Cause FERTILITY?")
debt_causes_fertility = grangercausalitytests(
    df_gc_diff[['Fertility', 'Debt']], 
    maxlag=max_lag, 
    verbose=True
)

print("\n" + "="*50 + "\n")

print(">> Testing: Does FERTILITY Granger-Cause DEBT?")
fertility_causes_debt = grangercausalitytests(
    df_gc_diff[['Debt', 'Fertility']], 
    maxlag=max_lag, 
    verbose=True
)
```

## 4. Välfärdsstatens paradox

Om fruktsamhetsnedgången driver skuldackumulering skulle sociala utgifter för hälsa, familjeförmåner och arbetslöshetsstöd kunna dämpa övergången. Men välfärdsstater expanderade just när fruktsamheten kollapsade i alla fyra länder som studeras här. Frågan är om de är kausalt kopplade, eller bara två symptom på samma djupare kraft.

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.stattools import grangercausalitytests
import warnings
warnings.filterwarnings('ignore')

# =========================================================================
# 1. PREPARE AND ALIGN BOTH LONG-RUN DATASETS
# =========================================================================
# Define the target names clearly
target_countries = ['United Kingdom', 'United States', 'Germany', 'France', 'Japan', 'Sweden', 'Italy']

# FIX: Filter by 'Country_name' instead of 'Code' so the strings match perfectly!
pivoted_fertility = df[df['Country_name'].isin(target_countries)].pivot(
    index='Year', 
    columns='Country_name', 
    values='TFR'
)

# Align the timelines based on available matching bounds
start_year = int(max(social_smoothed.index.min(), pivoted_fertility.index.min()))
end_year = int(min(social_smoothed.index.max(), pivoted_fertility.index.max()))
complete_years = np.arange(start_year, end_year + 1)

# Reindex and smooth fertility to create a continuous annual timeline
fertility_smoothed = pivoted_fertility.loc[start_year:end_year].reindex(complete_years).interpolate(method='linear')

print(f"Testtidslinje: {start_year} till {end_year}\n" + "="*60)

# =========================================================================
# 2. RUN GRANGER TEST PER COUNTRY
# =========================================================================
max_lag = 3 
country_to_test = 'United Kingdom' # This will now map flawlessly!

# Check which countries are actually available in BOTH datasets
available_fertility = set(fertility_smoothed.columns)
available_social = set(social_smoothed.columns)
common_countries = sorted(list(available_fertility & available_social))

print(f"Målländer: {target_countries}")
print(f"Tillgängligt i fertilitetsdata: {sorted(list(available_fertility))}")
print(f"Tillgängligt i sociala utgiftsdata: {sorted(list(available_social))}")
print(f"Gemensamma länder i BÅDA dataseten: {common_countries}\n")

# Pick a country that exists in both, falling back if the target isn't there
if country_to_test in common_countries:
    test_country = country_to_test
else:
    test_country = common_countries[0] if common_countries else None
    print(f"VARNING: '{country_to_test}' hittades inte i båda dataseten. Använder '{test_country}' istället.\n")

if test_country is None:
    raise ValueError("No common countries found between fertility and social spending datasets!")

# Build the temporary testing dataframe for the chosen country
df_country = pd.DataFrame({
    'Fertility': fertility_smoothed[test_country],
    'Social_Spending': social_smoothed[test_country]
}).dropna()

# Take the first difference to remove non-stationarity trends
df_country_diff = df_country.diff().dropna()

print(f"RESULTAT FÖR: {test_country.upper()}\n")
print(f">> Test 1: Granger-orsakar SOCIALA UTGIFTER FERTILITETSNEDGÅNG?")
res_social_causes_fert = grangercausalitytests(df_country_diff[['Fertility', 'Social_Spending']], maxlag=max_lag, verbose=False)
for lag in range(1, max_lag + 1):
    p_val = res_social_causes_fert[lag][0]['ssr_ftest'][1]
    print(f"   Lag {lag}: p-value = {p_val:.4f} " + ("*SIGNIFICANT*" if p_val < 0.05 else ""))

print("\n" + "-"*40)

print(f">> Test 2: Granger-orsakar FERTILITETSNEDGÅNG SOCIALA UTGIFTER?")
res_fert_causes_social = grangercausalitytests(df_country_diff[['Social_Spending', 'Fertility']], maxlag=max_lag, verbose=False)
for lag in range(1, max_lag + 1):
    p_val = res_fert_causes_social[lag][0]['ssr_ftest'][1]
    print(f"   Lag {lag}: p-value = {p_val:.4f} " + ("*SIGNIFICANT*" if p_val < 0.05 else ""))
```

Tänk på ett snurrande mynt. Vid hög hastighet håller det sig stadigt. Friktionen tömmer långsamt dess energi — årtionden av stadig fruktsamhetsnedgång. Så småningom når myntet en kritisk tröskel där det inte längre kan rätta upp sig efter små störningar. Det vacklar kraftigt, för att sedan lägga sig platt mot bordet. Detta är vad Ernest Hemingways Mike Campbell beskrev: "Hur gick du i konkurs? På två sätt. Långsamt, sedan plötsligt."

Det cybernetiska systemet saknar styrspakar eftersom senkapitalismens kärndirektiv (tillgångsprisskydd, kapitalrörlighet, aktieägarvärdemaximering) är oförenliga med de korrigeringar som krävs. Bostadsmarknaden kan inte krascha eftersom den stödjer banksystemet. Kapital kan inte omfördelas eftersom det kommer att fly. Algoritmiska flöden kan inte avoptimeras eftersom engagemangsmått driver intäkter.

Vi kan modellera vad systemet gör. Det som följer är en tillståndsrymddigitaltvilling som extraherar de dolda strukturella lager som driver varje lands trajektori.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# 1. Re-verify data alignment for Sweden
twin_country = 'Sweden'
twin_data = pd.DataFrame({
    'Fertility': fertility_smoothed.loc[1950:2020, twin_country],
    'Debt_to_GDP': debt_smoothed.loc[1950:2020, twin_country] if isinstance(debt_smoothed, pd.DataFrame) else debt_smoothed.loc[1950:2020],
    'Urbanization': urban_smoothed.loc[1950:2020, twin_country],
    'Education': edu_smoothed.loc[1950:2020, twin_country],
    'Child_Mortality': mort_smoothed.loc[1950:2020, twin_country]
}).dropna()

twin_data_scaled = (twin_data - twin_data.mean()) / twin_data.std()

# 2. Fit the state-space twin
model = sm.tsa.DynamicFactor(twin_data_scaled, k_factors=2, factor_order=1, error_order=1)
results = model.fit(disp=False, maxiter=250)

# =========================================================================
# 3. RUN THE WHAT-IF SIMULATION (Impulse Response Functions)
# =========================================================================
# We simulate a shock over a 20-year horizon
horizon = 20
irfs = results.impulse_responses(steps=horizon, impulse=0) # 0 index matches the first column: Fertility

print(f"Simulerar en positiv politikchock för fertilitet för {twin_country.upper()}...")

# =========================================================================
# 4. PLOT THE TRANSITORY SPREAD
# =========================================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 8), sharex=True)

# Panel 1: Hur chocken förändrar tvillingens dolda lager
# Slicing the factors from the impulse response matrix
ax1.plot(range(horizon + 1), irfs.iloc[:, 0], color='#800080', linestyle='--', linewidth=2, label='Hidden Layer 1: Modernization Shift')
ax1.plot(range(horizon + 1), irfs.iloc[:, 1], color='#008080', linewidth=2.5, label='Hidden Layer 2: Systemic Strain')
ax1.axhline(0, color='black', linestyle=':', alpha=0.5)
ax1.set_title("En tillfällig fruktsamhetschock avklingar inom 8 år utan att ändra den strukturella attraktorn", fontsize=12, fontweight='bold')
ax1.set_ylabel("Standard Deviation Change")
ax1.grid(True, linestyle=':', alpha=0.5)
ax1.legend()

# Panel 2: Hur chocken påverkar observerad makroskuld jämfört med baslinjen
# Mapping the specific named column for Debt from the output components
debt_col = [col for col in irfs.columns if 'Debt' in str(col)][0]
ax2.plot(range(horizon + 1), irfs[debt_col], color='#d62728', linewidth=2.5, label='Debt-to-GDP Trajectory')
ax2.axhline(0, color='black', linestyle=':', alpha=0.5)
ax2.set_title("Skuldsättning i förhållande till BNP sjunker tillfälligt men återgår till baslinjen inom 10 år", fontsize=12, fontweight='bold')
ax2.set_xlabel("Years Elapsed Since Policy Shock")
ax2.set_ylabel("Standard Deviation Change")
ax2.grid(True, linestyle=':', alpha=0.5)
ax2.legend()

plt.suptitle(f"Politiska ingripanden kan inte rubba systemet från dess strukturella trajektori ({twin_country.upper()})", fontsize=15, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tsa.api import VAR
import warnings
warnings.filterwarnings('ignore')

countries = ['Sweden', 'Japan', 'United States', 'Italy']
cybernetic_summary = []

print("Exekverar cybernetisk stabilitetsavsökning över digitala tvillingar...\n" + "="*75)

for country in countries:
    # 1. Sammanställ den synkroniserade datamatrisen
    df_country = pd.DataFrame({
        'Fertility': fertility_smoothed.loc[1950:2020, country],
        'Debt': debt_smoothed.loc[1950:2020, country] if isinstance(debt_smoothed, pd.DataFrame) else debt_smoothed.loc[1950:2020],
        'Urbanization': urban_smoothed.loc[1950:2020, country],
        'Education': edu_smoothed.loc[1950:2020, country],
        'Child_Mortality': mort_smoothed.loc[1950:2020, country]
    }).dropna()
    
    df_scaled = (df_country - df_country.mean()) / df_country.std()
    
    # 2. Extrahera dolda tillstånd från den dynamiska faktormotorn
    dfm = sm.tsa.DynamicFactor(df_scaled, k_factors=2, factor_order=1, error_order=1)
    res_dfm = dfm.fit(disp=False, maxiter=250)
    
    # Safely transpose the 2 latent factors
    latent_brain = pd.DataFrame(res_dfm.smoothed_state[:2, :].T, columns=['F1', 'F2'])
    
    # 3. Anpassa en VAR(1) direkt till den latenta hjärnan för att få övergångsmatris A
    var_engine = VAR(latent_brain)
    res_var = var_engine.fit(maxlags=1)
    A_matrix = res_var.coefs[0] # The 2x2 cybernetic transition matrix
    
    # 4. Beräkna egenvärdena för systemmatrisen
    eigenvalues = np.linalg.eigvals(A_matrix)
    spectral_radius = np.max(np.abs(eigenvalues))
    
    # 5. Beräkna systemets halveringstid (Dämpningshastighet)
    # Om spektralradie >= 1 är halveringstiden oändlig (explosiv loop)
    if spectral_radius < 1:
        half_life = np.log(0.5) / np.log(spectral_radius)
        status = "Homeostatic Attractor (Stable)"
    else:
        half_life = np.inf
        status = "Explosive Feedback (Unstable)"
        
    cybernetic_summary.append({
        'Country': country.upper(),
        'Eigenvalue_1': eigenvalues[0],
        'Eigenvalue_2': eigenvalues[1],
        'Spectral_Radius': spectral_radius,
        'System_Status': status,
        'Dampening_Half_Life_Years': round(half_life, 2) if half_life != np.inf else "Infinite"
    })

# =========================================================================
# SKRIV UT DEN CYBERNETISKA SAMMANSTÄLLNINGEN
# =========================================================================
df_cyber = pd.DataFrame(cybernetic_summary)
print(df_cyber.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

plt.close('all')
fig, axs = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle("Attraktorprognoser för 2050 bekräftar att alla banor konvergerar mot högre belastningsbassänger", fontsize=16, fontweight='bold')

axs_flat = axs.flatten()
colors = ['#800080', '#008080', '#1f77b4', '#d62728']
countries = ['Sweden', 'Japan', 'United States', 'Italy']

print("Calculating 2050 Steady-State Attractors using recursive system transitions...")

for i, country in enumerate(countries):
    ax = axs_flat[i]
    
    # 1. Compile and scale data
    df_c = pd.DataFrame({
        'Fertility': fertility_smoothed.loc[1950:2020, country],
        'Debt': debt_smoothed.loc[1950:2020, country] if isinstance(debt_smoothed, pd.DataFrame) else debt_smoothed.loc[1950:2020],
        'Urbanization': urban_smoothed.loc[1950:2020, country],
        'Education': edu_smoothed.loc[1950:2020, country],
        'Child_Mortality': mort_smoothed.loc[1950:2020, country]
    }).dropna()
    df_s = (df_c - df_c.mean()) / df_c.std()
    
    # 2. Fit Dynamic Factor Model
    m = sm.tsa.DynamicFactor(df_s, k_factors=2, factor_order=1, error_order=1)
    r = m.fit(disp=False, maxiter=250)
    
    # 3. Extract Historical Latent Factors (1950-2020)
    # Factor 1 and Factor 2 sit in the first two rows of the smoothed state matrix
    x_hist = r.smoothed_state[0, :]
    y_hist = r.smoothed_state[1, :]
    
    # =========================================================================
    # CYBERNETISK FAST LOOP: Projicera den fulla tillståndsvektorn ut till 2050
    # =========================================================================
    # Extrahera den sanna 5x5 systemövergångsmatrisen A och sluttillståndsvektorn från 2020
    A_matrix = r.filter_results.transition[:, :, 0]
    current_state = r.smoothed_state[:, -1]
    
    x_fore = []
    y_fore = []
    
    # Simulera 30 år in i framtiden (2020 -> 2050)
    for _ in range(30):
        current_state = A_matrix @ current_state
        x_fore.append(current_state[0]) # Future Factor 1
        y_fore.append(current_state[1]) # Future Factor 2
        
    x_fore = np.array(x_fore)
    y_fore = np.array(y_fore)
    
    # Kombinera historik- och prognosvektorer
    x_total = np.concatenate([x_hist, x_fore])
    y_total = np.concatenate([y_hist, y_fore])
    
    # =========================================================================
    # 5. GEOMETRIC PLOTTING
    # =========================================================================
    # Plot historical vector path
    ax.plot(x_total[:len(x_hist)], y_total[:len(y_hist)], color=colors[i], linewidth=2.5, alpha=0.8, label='Historisk bana (1950–2020)')
    # Plot projected future trajectory to 2050
    ax.plot(x_total[len(x_hist)-1:], y_total[len(y_hist)-1:], color=colors[i], linewidth=2, linestyle='--', alpha=0.9, label='Projicerad bana (2020–2050)')
    
    # Strukturella milstolpar
    ax.scatter(x_total[0], y_total[0], color='green', s=120, zorder=5, label='1950 – Ursprung')
    ax.scatter(x_hist[-1], y_hist[-1], color='orange', s=100, zorder=5, label='2020 – Vektor')
    ax.scatter(x_total[-1], y_total[-1], color='red', s=150, marker='X', zorder=6, label='2050 – Attraktormål')
    
    ax.set_title(f"Attraktorhorisont – {country.upper()}", fontsize=12, fontweight='bold')
    ax.set_xlabel("Factor 1: Socio-Economic Shift")
    ax.set_ylabel("Factor 2: Systemic Strain")
    ax.grid(True, linestyle=':', alpha=0.5)
    ax.legend(loc='best')

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

```python
import json
import pickle
import os
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Global payload dictionary for micro-API routing
cyber_api_payload = {}
os.makedirs("../../artifacts/notebook", exist_ok=True)
countries = ['Sweden', 'Japan', 'United States', 'Italy']

print("Initierar serialiseringssekvens för cybernetiska tvillingar...")

for country in countries:
    # 1. Compile identical historical baselines
    df_c = pd.DataFrame({
        'Fertility': fertility_smoothed.loc[1950:2020, country],
        'Debt': debt_smoothed.loc[1950:2020, country] if isinstance(debt_smoothed, pd.DataFrame) else debt_smoothed.loc[1950:2020],
        'Urbanization': urban_smoothed.loc[1950:2020, country],
        'Education': edu_smoothed.loc[1950:2020, country],
        'Child_Mortality': mort_smoothed.loc[1950:2020, country]
    }).dropna()
    
    # Capture scaling parameters required to de-standardize future API predictions
    means = df_c.mean().to_dict()
    stds = df_c.std().to_dict()
    
    df_s = (df_c - df_c.mean()) / df_c.std()
    
    # 2. Fit the finalized Dynamic Factor framework
    m = sm.tsa.DynamicFactor(df_s, k_factors=2, factor_order=1, error_order=1)
    r = m.fit(disp=False, maxiter=250)
    
    # 3. Extract the physical matrix structures
    # A_matrix represents the full internal state transition equations (5x5 matrix)
    A_matrix = r.filter_results.transition[:, :, 0]
    state_2020 = r.smoothed_state[:, -1]
    
    # 4. Compute explicit 2050 terminal target coordinates
    current_state = state_2020.copy()
    for _ in range(30):
        current_state = A_matrix @ current_state
        
    # 5. Fyll den lätta JSON-ordboken
    cyber_api_payload[country.lower()] = {
        'transition_matrix_A': A_matrix.tolist(),
        'latent_state_2020': state_2020.tolist(),
        'latent_attractor_2050': current_state.tolist(),
        'normalization_means': means,
        'normalization_stds': stds
    }
    
    # 6. Exportera fulla statistiska modellbinärer via Pickle
    pickle_filename = f"../../artifacts/notebook/twin_model_{country.lower()}.pkl"
    with open(pickle_filename, 'wb') as f:
        pickle.dump(r, f)
    print(f"   Sparade fullt objekt-binär -> {pickle_filename}")

# 7. Exportera den lätta parametermatrisen via JSON
json_filename = "../../artifacts/notebook/cybernetic_twins_api_payload.json"
with open(json_filename, 'w') as f:
    json.dump(cyber_api_payload, f, indent=4)

print("\n" + "="*75)
print("LYCKAT: Micro-API-blåkopian är kompilerad och fryst för produktion.")
print(f"JSON-parametermanifest: '{json_filename}'")
print("="*75)
```

## Epilog: De fyra cybernetiska botemedlen

Om ett cybernetiskt system är instängt i eskalerande positiv återkoppling identifierade fältets grundare (Ashby, Wiener, Beer) fyra möjliga vägar:

1. **Metasystemsövergång**: En högre nivås kontroller skriver om de lägre systemens regler. Förbud mot bostadsfinansialisering. Reglering av algoritmiska flöden.

2. **Återställande av regulatorn**: Återkoppling av frikopplade återkopplingsloopar. Knyta pensionsutbetalningar till antalet uppfostrade barn, inte bara livsinkomst.

3. **Nödvändig variation (Requisite Variety)**: Regulatorn måste ha åtminstone lika mycket komplexitet som systemet den styr. Anpassningsbara arbetslivsarkitekturer. Valutor som förlorar värde över tid för att motverka tillgångshamstring.

4. **Fasövergång**: Låta den positiva återkopplingsloopen rulla tills strukturutmattning tvingar fram en återställning. Myntet faller. Systemet avlägsnar sin ohållbara komplexitet och stabiliserar sig i en enklare attraktorbassäng.

Alla fyra vägar är synliga. Ingen tas för närvarande.

```python
import json
import numpy as np

def run_cybernetic_forecast(country_name, steps_ahead=10):
    # Ladda de lätta matrisparametrarna
    with open("../../artifacts/notebook/cybernetic_twins_api_payload.json", "r") as f:
        twin_config = json.load(f)[country_name.lower()]
        
    A = np.array(twin_config['transition_matrix_A'])
    x = np.array(twin_config['latent_state_2020'])
    
    # Iterera systemmatrisen genom tillståndsrymden
    for _ in range(steps_ahead):
        x = A @ x
        
    # Returnera de latenta tillståndsvektorernas koordinater för den prognoserade horisonten
    return {
        'forecasted_factor_1_modernization': x[0],
        'forecasted_factor_2_systemic_strain': x[1]
    }

# Omedelbar mikrosekundsexekvering
target_vector_2030 = run_cybernetic_forecast('sweden', steps_ahead=10)
print("Projicerade koordinater för 2030:", target_vector_2030)
```
